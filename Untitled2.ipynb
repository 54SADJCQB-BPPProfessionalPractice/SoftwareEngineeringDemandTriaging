#Exploratory Data Analysis (EDA): Converting Date Fields

Within this quick dataframe view of the top of the Demand Excel, it is noticed that the 'Submitted_Date' aand 'Target_Date' column uses YYYY-MM-DD and HH-MM-SS which isn't what the calculations need.

Tgriughout the calculations, errors='coerce' is used to ensure that invalid date values are converted to missing values rather then causing the program to fail. This is useful in devlopmet because the dataset consist of users inputs and therefore very prone to invalid values being pulled through - so it is importnat the cod eis robust enough to handle them

```python
df['Submitted_Date'] = pd.to_datetime(df['Submitted_Date'], errors='coerce').dt.strftime('%Y-%m-%d')
df.head()
```

As you can see now from cell 9 to 12 the 'Submitted_Date' has been changed to just YYYY-MM-DD, this also now need repeating for 'Target_Date'

```python
df['Target_Date'] = pd.to_datetime(df['Target_Date'], errors='coerce').dt.strftime('%Y-%m-%d')
df.head()
```

##Feature Engineering: Calculating the avaliable Delivery Time

Now we have cleaned up the Submitted and Target Date in our df, it would be useful to see how many days are left until our target date as this will help when triaging the most important work on a time-basis.

Therefore, a new feature called 'days_to_Target' is created by calculating the number of days between the submitted date and the requested target date

This provudes a numerical  measure of the timeframe avaliable for each demand request. Negative values are retained because they may indicate data quality issue where the target date is earlier then the submitted date

```python
df['Submitted_Date'] = pd.to_datetime(df['Submitted_Date'], errors='coerce')
df['Target_Date'] = pd.to_datetime(df['Target_Date'], errors='coerce')
df['Days_Until_Target'] = (df['Target_Date'] - df['Submitted_Date']).dt.days
df.head()
```

## Feature Engineering: Comparing Avaliable Time (Days_Until_Target) with Estimated Effort

The avaliable delivery timeframe is comared with 'Estimated_Effort_Days' to provide an indication of whether estimated work could fit within the requested timeframe, as this will aid with the Truaging criteria to ensure it is deliverable within the proposed timeframe.

Ensuring to priortise tighter timeframes over longer ones with higher criticality where justifiable

Therefore, a numerical feature called 'Margin_To_Delivery_Days' is created by subtracting the estimated effort from the number of days avaliable

A positive value = Number of Days remaining
Zero = An exact match
Negative value = Shows how many days the estimated effort exceeds the avaliable timeframe

```python
#Convert Estimated_Effort_Days to numeric first
df['Estimated_Effort_Days'] = pd.to_numeric(df['Estimated_Effort_Days'], errors='coerce')

#Compare the available timeframe with estimated effort
df['Margin_To_Delivery_Days'] = df['Days_Until_Target'] - df['Estimated_Effort_Days']

#Review the results (also fixing the column name from 'Days_To_Target' to 'Days_Until_Target')
df[['Demand_ID', 'Days_Until_Target', 'Estimated_Effort_Days', 'Margin_To_Delivery_Days']].head(10)
```
